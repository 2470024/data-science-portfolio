## データ加工・モデリング

変数の加工

```python
##### データの読み込み #####

# "tomato_data.csv" を作業フォルダにアップロードしたのち以下を実行

import pandas as pd

df_data_1 = pd.read_csv("../データ/tomato_data.csv")
df_data_1.head()
```

```python
## 欠損値があるカラムとその個数を表示

# 各カラム毎に欠損値(isnull)の数(sum)を確認
nullcounts_columns = df_data_1.isnull().sum()
display(nullcounts_columns)

# 欠損値が1つでもあるカラムと個数を表示
nullcounts_columns[nullcounts_columns>0]
```

```python
## 欠損値あるカラムのデータの入り方を確認
# 基本統計量を表示
display(df_data_1[['FruitsSugarLevel', 'FruitsAcidLevel', 'Soil_EC', 'Soil_PH']].describe().T)

# ヒストグラム
df_data_1[['FruitsSugarLevel', 'FruitsAcidLevel', 'Soil_EC', 'Soil_PH']].hist(bins=9,figsize=(12,6))
plt.show()
```

```python
##### データ加工：前処理 #####

## 日付データを対象に日付情報に加工
target = ['Date', 'SunriseTime', 'SunsetTime']
for t in target:
    df_data_1[t] = pd.to_datetime(df_data_1[t])

## 欠損値あるカラムを参考に何で埋めるかを決める（今回は0）
df_data_1 = df_data_1.fillna(int(0))

df_data_1.info()
```

```python
### データ加工：目的変数の加工 ###

import matplotlib.pyplot as plt
import math

# 階級数=3のヒストグラムを表示
plt_1 = plt.hist(df_data_1['Photosynthetic_Rate'],bins=3)
print(plt_1)
plt.xlabel('Photosynthetic_Rate')
plt.show()
```

```python
# 目的変数：純光合成量を3つのクラスに分けてカラムを追加
df_data_1.loc[df_data_1['Photosynthetic_Rate']<=3.49, 'Photosynthetic_class'] = "1"

df_data_1.loc[(df_data_1['Photosynthetic_Rate']>3.49)
              &(df_data_1['Photosynthetic_Rate']<=6.22), 'Photosynthetic_class'] = "2"

df_data_1.loc[df_data_1['Photosynthetic_Rate']>6.22, 'Photosynthetic_class'] = "3"

df_data_1.head()
```

```python
# 目的変数：翌日の純光合成量をカラムに追加
# (shiftメソッドでずらすことが可能。+1の場合下の行にずらし、-1の場合は上の行にずらす)
df_data_1['Photosynthetic_class_n1'] = df_data_1['Photosynthetic_class'].shift(-1)


# カラム"Photosynthetic_class_n1"の最後の行がnullになっていることが確認できる。
display(df_data_1.head(),df_data_1.tail())
```

```python
### データ加工・説明変数の加工
# 説明変数：平均値・差分などの変数をカラムに追加

# 昼夜平均CO2濃度
df_data_1["CO2_Mean"] = (df_data_1["CO2_day"] + df_data_1["CO2_night"])/2.0

# 昼夜平均湿度
df_data_1["RH_Mean"] = (df_data_1["RH_day"] + df_data_1["RH_night"])/2.0

# 昼夜平均飽差
df_data_1["HD_Mean"] = (df_data_1["HD_day"] + df_data_1["HD_night"])/2.0

# 昼夜平均室温
df_data_1["GHTemp_Mean"] = (df_data_1["GHTemp_day"] + df_data_1["GHTemp_night"])/2.0

# 昼夜室温差
df_data_1["GHTemp_Diff"] = df_data_1["GHTemp_day"] - df_data_1["GHTemp_night"]

# 土壌含水量
df_data_1["Uptake_Water"] = df_data_1["Drip_Water"] - df_data_1["Drain_Water"]

# 日長時間(分)
df_data_1['DayLength'] = ( ( df_data_1['SunsetTime'].dt.hour*60 + df_data_1['SunsetTime'].dt.minute) \
                                                 - ( df_data_1['SunriseTime'].dt.hour*60 + df_data_1['SunriseTime'].dt.minute) )

# 平年比日射量%
df_data_1["Radiation_Percent"] = df_data_1["Radiation"] / df_data_1["NormalRadiation"]

df_data_1.head()
```

```python
# 目的変数(翌日の純光合成量)が作成されない(=欠損値となる)最後の行を削除
df_data_1 = df_data_1.dropna(how='any')


# 日出時刻・日没時刻のデータ型をHH:MMの文字列に変換
df_data_1['SunriseTime'] = df_data_1['SunriseTime'].dt.strftime('%H:%M')
df_data_1['SunsetTime'] = df_data_1['SunsetTime'].dt.strftime('%H:%M')

df_data_1.head()
```

```python
### データの書き出し ###

# 日付順に並び替える
df_data_1 = df_data_1.sort_values('Date')

# 書き出し
df_data_1.to_csv("../データ/tomato_data_update.csv",index=False)
```

```python
### モデルの構築（Step1）：データセットの分割 ###

import datetime as dt
from sklearn.model_selection import train_test_split

# 空のデータフレームを作成
df_train_1 = pd.DataFrame()
df_test_1 = pd.DataFrame()


# 1月-12月まで繰り返し処理
for i in range(1,13):
    # 該当の月のデータのみを抽出
    df_m_1 = df_data_1[(df_data_1['Date'].dt.month == i)]
    # 学習用とテスト用にデータセットを分割
    df_train_m_1, df_test_m_1 = train_test_split(df_m_1, test_size=0.4, random_state=0)
    # 月ごとのデータをそれぞれ学習用とテスト用のデータセットとして統合
    df_train_1 = pd.concat([df_train_1, df_train_m_1], axis=0)
    df_test_1 = pd.concat([df_test_1, df_test_m_1], axis=0)

#学習用とテスト用のフラグを追加
df_train_1['data_flag'] = 'train'
df_test_1['data_flag'] = 'test'
df_data_1 = pd.concat([df_train_1, df_test_1], axis=0)
df_data_1 = df_data_1.sort_values('Date')

df_data_1.head()
```

```python
### ディープラーニング　ハイパラチューニング自動化 ###
import numpy as np
import timeit
from sklearn import datasets, model_selection, metrics
from sklearn.neural_network import MLPClassifier

# ハイパーパラメータのリスト
tuned_parameters = [
    {
        'max_iter': [100],
        'activation': ['identity', 'logistic', 'tanh', 'relu'],
        'solver': ['lbfgs', 'sgd', 'adam'],
        'hidden_layer_sizes': [\
                              (100),
                              (100,10),
                              (100,100,10),
                              (100,100,100,10),
                              (100,100,100,100,10)
                              ]
    }
]

# 分類器とハイパーパラメータのリストを定義
licv_1 = model_selection.GridSearchCV(MLPClassifier(
    early_stopping = True),
    param_grid = tuned_parameters,
    scoring = 'accuracy',
    cv = 5
)
```
